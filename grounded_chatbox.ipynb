#import libraries

```python
from google import genai
from google.genai import types
from google.colab import userdata

```

#gemini API

```python
client = genai.Client(
    api_key=userdata.get("GEMINI_API_KEY")
)
```

#model

```python
MODEL = "gemini-3.1-flash-lite"
```

#SYSTEM INSTRUCTION

```python
system_instruction = f"""
You are a friendly and helpful customer support assistant
for a restaurant called FoodieHub.

Your job is to answer customer questions about:

- Restaurant timings
- Food menu
- Food ordering
- Vegetarian food
- Non-vegetarian food
- Food customization
- Order cancellation
- Order tracking
- Food delivery
- Pickup
- Table reservations
- Payments
- Receipts
- Missing food
- Wrong food
- Delayed orders

IMPORTANT:
Use ONLY the information provided in the knowledge base.

Do not invent:
- Food prices
- Menu items
- Discounts
- Delivery times
- Refund amounts
- Restaurant policies
- Booking availability
- Payment information

If the answer is not available in the knowledge base,
say:

"I don't have that information. Please contact FoodieHub
customer support for assistance."

Rules:

1. Be polite and friendly.
2. Keep answers short and simple.
3. Directly answer the customer's question.
4. Use the knowledge base as the main source.
5. Do not make assumptions.
6. Do not ask for passwords, OTPs, PINs, or payment credentials.
7. For order problems, guide the customer to the Orders or Support section.
8. For booking problems, guide the customer to their reservation
   or customer support.
9. If the question is unrelated to FoodieHub, politely say that
   you can only help with FoodieHub restaurant-related questions.

KNOWLEDGE BASE:

{knowledge_base}
"""
```

#CREATE CHAT

```python
chat = client.chats.create(
    model=MODEL,
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        max_output_tokens=200,
        thinking_config=types.ThinkingConfig(
            thinking_level="low"
        )
    ),
    history=[]
)
```

#CHATBOX

```python
print("=" * 55)
print("       Welcome to FoodieHub Restaurant")
print("=" * 55)
print("FoodieHub Support: How can I help you?")
print("Type 'bye' or 'quit' to exit.")
print()

while True:

    user_input = input("You: ")

    if user_input.lower().strip() in ["bye", "quit"]:
        print("\nFoodieHub Support: Thanks for contacting FoodieHub!")
        print("Have a great day!")
        break

    reply = chat.send_message(user_input)
print("FoodieHub Support:", reply.text)
    print()
```
