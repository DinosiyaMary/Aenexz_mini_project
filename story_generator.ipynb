```python
from google import genai
from google.genai import types
from google.colab import userdata
```

#gemini API

```python
client = genai.Client(
    api_key=userdata.get("GEMINI_API_KEY")
)
```

#model

```python
MODEL = "models/gemini-2.5-flash-lite"
```

```python
#RETAURANT GENERTOR FUNCTION
```

```python
def generate_restaurant(restaurant_name, cuisine, occasion, details):

    system_prompt_content = """
You are a professional restaurant assistant.

Your task is to create a useful and attractive restaurant
description based on the information provided by the user.

Follow these instructions:

- Create an engaging restaurant description.
- Mention the restaurant name.
- Mention the cuisine.
- Mention the occasion.
- Include the important details provided by the user.
- Describe the dining experience naturally.
- Make the content attractive and easy to understand.
- Do not invent specific prices, addresses, phone numbers,
  opening hours, or menu items unless the user provides them.
- Keep the information consistent with the user's input.
- Output only the restaurant description.
"""

    user_prompt = f"""
Create a restaurant description.

Restaurant Name: {restaurant_name}
Cuisine: {cuisine}
Occasion: {occasion}
Restaurant Details: {details}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=user_prompt,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt_content,
            temperature=1,
            max_output_tokens=500
        )
    )

    print(response.text)
```

#USER INPUT

```python
print("=" * 60)
print("             RESTAURANT GENERATOR")
print("=" * 60)

restaurant_name = input("Enter restaurant name: ")
cuisine = input("Enter cuisine type: ")
occasion = input("Enter occasion: ")
details = input("Enter restaurant details: ")

print("\nGenerated Restaurant Description:\n")

generate_restaurant(
    restaurant_name,
    cuisine,
    occasion,
    details
)
```
